```python
import chromadb

client = chromadb.Client()
collection = client.create_collection(name="my_collection", get_or_create=True)
```

```python
collection .add (
    documents = [
        "This document is about New York",
        "This document is about Delhi"
    ],
    ids = ['id1', 'id2']
)
```

```python
all_docs = collection.get()
all_docs
```
